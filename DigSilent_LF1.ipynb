# Checking if the Python is correctly connecting with DigSilent or Not

```python
import sys
pf_path = r"C:\Program Files\DIgSILENT\PowerFactory 2024\Python\3.11"
if pf_path not in sys.path:
    sys.path.append(pf_path)
try:
    import powerfactory
except ImportError:
    print("Error: Could not import powerfactory module. Please check the path.")
    sys.exit(1)
```

## Creating a function to check the success or failure of the connection:
Connect to PowerFactory Engine Mode.\
Returns\
app : PowerFactory Application object \
Display the output as success or failure of the connection

```python
def connect_pf():
    app = powerfactory.GetApplicationExt()
    if app is None:
        raise RuntimeError("Unable to connect to PowerFactory.")
    return app

print("=" * 60)
print("POWERFACTORY CONNECTION TEST")
print("=" * 60)

try:
    app = connect_pf()
    print("Status : Connected Successfully")
    print("Application :", app)
    print("=" * 60)
except Exception as err:
    print("Status : Connection Failed")
    print(err)
    print("=" * 60)
```

## When the Python connects successfully with the Digsilent PF, its time to show the following: 
- User name
- Active Project - if there is any before activating a project through python code
- Study Case  - if there is any before activating through python code


```python
try:
    user = app.GetCurrentUser() # gets the user name of the DigSILENT PowerFactory application
    if user:
     print(f"Current User    : {user.loc_name}")
    else:
     print("Current User    : None")

except Exception:
    print("Current User    : Unable to retrieve")

try:
    project = app.GetActiveProject() # gets the active project in the DigSILENT PowerFactory application
    if project:
       print(f"Active Project  : {project.loc_name}")
    else:
       print("Active Project  : None")

except Exception: #tell the user that the active project could not be retrieved
            print("Active Project  : Unable to retrieve")


try:
    studycase = app.GetActiveStudyCase() # gets the active study case in the DigSILENT PowerFactory application
    if studycase:
        print(f"Study Case      : {studycase.loc_name}")
    else:
        print("Study Case      : None")

except Exception: #indicates that the active study case could not be retrieved
            print("Study Case      : Unable to retrieve")

print("=" * 60)
```
